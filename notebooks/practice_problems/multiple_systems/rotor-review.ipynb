## Quest 2: The $Z$-axis rotation
**Goal:** Understand what $R_Z(\theta())$ does to a quantum state.
**Setup:** 
    - Start with $$\ket{\psi}=\frac{1}{\sqrt{2}}\begin{bmatrix}1 \\ 1\end{bmatrix} \, .$$
    - Apply the rotation $$R_Z(\theta)\ket{\psi}=\frac{1}{\sqrt{2}}\begin{bmatrix}e^{-i\theta / 2} \\ e^{i\theta / 2}\end{bmatrix} \, .$$

**Tasks:** For each angle $$\theta \in \{ 0, \frac{\pi}{2}, \pi, 2\pi \}$$
    a) Calculate the resulting state.
    b) Calculate the measurement probabilities.
    c) Compare the resulting state with the present state.
    d) Determine wihether teh difference is physcially observable or merely a global phase.

**Important Observation:** $$R_Z(\theta)\ket{0} = e^{-i\theta / 2}\ket{0}$$
- This state has acquired a phase, but its measured probablities haven't changed. The same is true for $\ket{1}$.
- However, the equal superposition state behaves differently: its two components acquire different phases. This relative phase is significant.
- **Challenge:** Change the angle and observe what happens to the amplitudes. Why do the probabilities remain equal?
