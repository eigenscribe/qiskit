# Partial Measurements
Recall: 
- We can simulate the measurement of a quantum state using the `meausre` method.
    - Returns two results: 
        1. simulated measurement result
        2. updated state vector
- We can decalre a quantum state vector using the `Statevector` command

```python
from qiskit.quantum_info import Statevector, Operator
from numpy import sqrt
from IPython.display import display, Latex, Math

from qiskit import __version__
print(__version__)
```

## 🔵 Example:
Create the state $$\ket{w} = \frac{\ket{001} + \ket{010} + \ket{100}}{\sqrt{3}}$$
and measure qubit number 0 (rightmost qubit).

```python
w = Statevector([0, 1, 1, 0, 1, 0, 0, 0] / sqrt(3))
print('Initial state:')
display(w.draw("latex"))

result, state = w.measure([0])      # measure qubit zero
print(f'Measurement outcome of 0th qubit: {result}')
print('Final state:')
display(state.draw("latex"))        # display new qubit state
```

```python
w = Statevector([0, 1, 1, 0, 1, 0, 0, 0] / sqrt(3))
display(w.draw("latex"))
 
result, state = w.measure([0])
print(f"Measured: {result}\nState after measurement:")
display(state.draw("latex"))
 
result, state = w.measure([0, 1])
print(f"Measured: {result}\nState after measurement:")
display(state.draw("latex"))
```
